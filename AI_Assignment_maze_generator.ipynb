## Import libraries

```python
import numpy as np
import random
```

## Constants

```python
# Constants
GRID_SIZE = 150
```

## Maze Functions

```python
class DisjointSet:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, v):
        if self.parent[v] != v:
            self.parent[v] = self.find(self.parent[v])
        return self.parent[v]

    def union(self, u, v):
        root_u = self.find(u)
        root_v = self.find(v)
        if root_u != root_v:
            if self.rank[root_u] > self.rank[root_v]:
                self.parent[root_v] = root_u
            elif self.rank[root_u] < self.rank[root_v]:
                self.parent[root_u] = root_v
            else:
                self.parent[root_v] = root_u
                self.rank[root_u] += 1
            return True
        return False
```

```python
def create_maze(size=GRID_SIZE):
    num_cells = size * size
    edges = []  

    def cell_index(row, col):
        return row * size + col
    
    for r in range(size):
        for c in range(size):
            if c < size - 1:  # Right neighbor
                edges.append((random.random(), cell_index(r, c), cell_index(r, c + 1)))
            if r < size - 1:  # Bottom neighbor
                edges.append((random.random(), cell_index(r, c), cell_index(r + 1, c)))

    random.shuffle(edges)
    
    dsu = DisjointSet(num_cells)
    maze = [[1] * (2 * size + 1) for _ in range(2 * size + 1)]  

    for _, u, v in edges:
        if dsu.union(u, v):
            ru, cu = divmod(u, size)
            rv, cv = divmod(v, size)

            maze[2 * ru + 1][2 * cu + 1] = 0  
            maze[2 * rv + 1][2 * cv + 1] = 0  
            maze[(2 * ru + 1 + 2 * rv + 1) // 2][(2 * cu + 1 + 2 * cv + 1) // 2] = 0  

    maze[1][1] = 0  
    maze[-2][-2] = 0  

    # Add extra paths to introduce multiple routes
    add_extra_paths(maze, size)

    return maze
```

```python
def add_extra_paths(maze, size):
    num_extra_paths = int(0.30 * (size * size))
    possible_walls = []

    for r in range(1, len(maze) - 1, 2):
        for c in range(1, len(maze[0]) - 1, 2):
            if maze[r][c] == 0:
                if c + 2 < len(maze[0]) - 1 and maze[r][c + 2] == 0 and maze[r][c + 1] == 1:
                    possible_walls.append((r, c + 1))
                if r + 2 < len(maze) - 1 and maze[r + 2][c] == 0 and maze[r + 1][c] == 1:
                    possible_walls.append((r + 1, c))

    random.shuffle(possible_walls)
    for _ in range(num_extra_paths):
        if possible_walls:
            r, c = possible_walls.pop()
            maze[r][c] = 0
```

## Main Function

```python
def main():
    # Create maze
    maze = create_maze(GRID_SIZE)
    
    np.save(f"./Mazes/{GRID_SIZE}.npy", np.array(maze))
    
    print("Maze generated and stored for size: ", GRID_SIZE)
```

```python

if __name__ == "__main__":
    main()
```
